```python
import ffmpeg
import os

def extract_audio(video_path, audio_path):
    if not os.path.exists(video_path):
        raise FileNotFoundError(f"Файл {video_path} не найден. Пожалуйста, загрузите его через панель выше.")

    if os.path.exists(audio_path):
        os.remove(audio_path)

    print(f"Обработка {video_path}...")
    try:
        # Извлекаем аудио дорожку и сохраняем в wav
        ffmpeg.input(video_path).output(audio_path, acodec='pcm_s16le', ac=1, ar='16k').run(overwrite_output=True, capture_stdout=True, capture_stderr=True)
        return True
    except ffmpeg.Error as e:
        print(f"Ошибка FFmpeg: {e.stderr.decode()}")
        return False

# Запуск процесса
print("--- Шаг: Извлечение аудио ---")
try:
    if extract_audio(INPUT_VIDEO, "audio.wav"):
        print("✅ Успех: Файл 'audio.wav' готов к сегментации.")
    else:
        print("❌ Не удалось извлечь аудио.")
except FileNotFoundError as e:
    print(f"⚠️ Внимание: {e}")
except Exception as e:
    print(f"🧨 Непредвиденная ошибка: {e}")
```

```python
from pydub import AudioSegment
from pydub.silence import split_on_silence
import os

def segment_audio(audio_path, output_dir):
    if not os.path.exists(audio_path):
        raise FileNotFoundError(f"File {audio_path} not found. Please run the audio extraction step first.")

    # Load audio
    audio = AudioSegment.from_wav(audio_path)

    # Split based on silence (min_silence_len in ms, silence_thresh in dBFS)
    print("Analyzing silence and splitting...")
    chunks = split_on_silence(
        audio,
        min_silence_len=1000,
        silence_thresh=audio.dBFS - 14,
        keep_silence=500
    )

    segments_info = []
    for i, chunk in enumerate(chunks):
        output_filename = f"segment_{i:03d}.wav"
        output_path = os.path.join(output_dir, output_filename)
        chunk.export(output_path, format="wav")

        segments_info.append({
            'id': i,
            'path': output_path,
            'duration': len(chunk) / 1000.0
        })

    return segments_info

# Сегментируем аудио
print("Сегментируем аудио по паузам...")
try:
    segments = segment_audio("audio.wav", SEGMENTS_DIR)
    print(f"Создано {len(segments)} сегментов для обработки")

    # Показываем информацию о сегментах
    if segments:
        print("\nПримеры созданных сегментов:")
        for seg in segments[:5]:  # показываем первые 5 сегментов
            print(f"Сегмент {seg['id']}: {seg['duration']:.2f} с, файл: {seg['path']}")
    else:
        print("Тишина не найдена или файл слишком короткий для сегментации.")
except Exception as e:
    print(f"Ошибка при сегментации: {e}")
```

```python
import whisper_timestamped as whisper
import json

class ASTPipeline:
    def __init__(self, model_name="base"):
        print(f"Загрузка модели Whisper ({model_name})...")
        self.model = whisper.load_model(model_name, device="cpu")

    def translate_segment(self, audio_path):
        # В данном контексте Whisper используется для распознавания.
        # Для перевода на другой язык можно добавить параметр translation.
        audio = whisper.load_audio(audio_path)
        result = whisper.transcribe(self.model, audio, language="ru")
        return result["text"].strip()

# Инициализируем AST pipeline
try:
    if 'segments' in locals() and segments:
        print("Инициализируем Whisper для обработки сегментов...")
        ast_pipeline = ASTPipeline()

        # Обрабатываем все сегменты
        translated_texts = []
        print("Начинаем обработку сегментов...")

        for i, seg in enumerate(segments):
            print(f"Обрабатываем сегмент {i+1}/{len(segments)}: {seg['path']}")
            text = ast_pipeline.translate_segment(seg["path"])
            translated_texts.append(text)
            print(f"Текст: {text[:100]}...")

        print(f"\n✅ Все сегменты ({len(segments)}) успешно обработаны!")
    else:
        print("⚠️ Список сегментов пуст. Пожалуйста, запустите ячейку сегментации выше.")
except Exception as e:
    print(f"❌ Ошибка при транскрибации: {e}")
```

```python
from gtts import gTTS
import os

class TTSPipeline:
    def __init__(self):
        print("TTS Pipeline инициализирован.")

    def synthesize(self, text, output_path, target_duration=None):
        tts = gTTS(text=text, lang='ru')
        tts.save(output_path)
        return output_path

# Инициализируем TTS pipeline
print("Подготовка к синтезу речи...")
tts_pipeline = TTSPipeline()

# Проверяем наличие необходимых данных из предыдущих шагов
if 'segments' in locals() and 'translated_texts' in locals():
    tts_paths = []
    print(f"Начинаем синтез речи для {len(segments)} сегментов...")

    for i, (seg, text) in enumerate(zip(segments, translated_texts)):
        if not text.strip():
            print(f"Пропуск пустого сегмента {i+1}")
            tts_paths.append(None)
            continue

        output_path = os.path.join(TTS_DIR, f"tts_{i:03d}.wav")
        try:
            tts_path = tts_pipeline.synthesize(text, output_path, seg["duration"])
            tts_paths.append(tts_path)
            print(f"Синтезирован сегмент {i+1}/{len(segments)}")
        except Exception as e:
            print(f"Ошибка на сегменте {i}: {e}")
            tts_paths.append(None)

    print(f"\n✅ Готово! Файлы в папке: {TTS_DIR}")
else:
    print("⚠️ Ошибка: Переменные 'segments' или 'translated_texts' не найдены. Сначала выполните перевод (ячейка 4vdtLj-oMS2G).")
```

```python
from pydub import AudioSegment
import os

def mix_audio_segments(paths, output_filename):
    """
    Concatenates audio segments from the provided list of paths.
    """
    valid_paths = [p for p in paths if p and os.path.exists(p)]
    if not valid_paths:
        print("❌ Нет доступных аудиофайлов для объединения.")
        return

    combined = AudioSegment.empty()
    for path in valid_paths:
        segment_audio = AudioSegment.from_wav(path)
        combined += segment_audio

    combined.export(output_filename, format="wav")
    print(f"✅ Файл успешно сохранен: {output_filename}")

# Запуск объединения
if 'tts_paths' in locals() and tts_paths:
    print("Объединяем аудиосегменты...")
    mix_audio_segments(tts_paths, "final_audio.wav")
else:
    print("⚠️ Список 'tts_paths' пуст. Пожалуйста, выполните синтез речи (ячейка 2773b01a) сначала.")
```

```python
import subprocess

def assemble_final_video(video_input, tts_paths, segments, output_video):
    # Создаем файл со списком для ffmpeg concat
    concat_file = "concat_list.txt"
    with open(concat_file, "w") as f:
        for path in tts_paths:
            if path:
                f.write(f"file '{path}'\n")

    # Склеиваем аудио
    print("Склеивание аудио сегментов...")
    subprocess.run(["ffmpeg", "-y", "-f", "concat", "-safe", "0", "-i", concat_file, "-c", "copy", "temp_audio.wav"], capture_output=True)

    # Накладываем новое аудио на оригинальное видео
    print("Создание финального видео...")
    subprocess.run([
        "ffmpeg", "-y", "-i", video_input, "-i", "temp_audio.wav",
        "-c:v", "copy", "-map", "0:v:0", "-map", "1:a:0", "-shortest", output_video
    ], capture_output=True)

    print(f"✅ Готово! Результат сохранен в: {output_video}")

if 'tts_paths' in locals() and any(tts_paths):
    assemble_final_video(INPUT_VIDEO, tts_paths, segments, "final_dubbed_video.mp4")
else:
    print("⚠️ Сначала нужно успешно выполнить синтез речи (TTS).")
```

```python
import subprocess
import os

OUTPUT_VIDEO = "final_dubbed_video.mp4"

def replace_audio_in_video(video_input, audio_input, video_output):
    if not os.path.exists(video_input):
        print(f"❌ Ошибка: Входное видео {video_input} не найдено.")
        return
    if not os.path.exists(audio_input):
        print(f"❌ Ошибка: Аудиофайл {audio_input} не найден. Сначала выполните объединение.")
        return

    print(f"Склеиваем {video_input} и {audio_input}...")
    try:
        # -map 0:v:0 берет видео из первого файла, -map 1:a:0 берет аудио из второго
        # -shortest обрезает по самому короткому потоку
        subprocess.run([
            "ffmpeg", "-y", "-i", video_input, "-i", audio_input,
            "-c:v", "copy", "-map", "0:v:0", "-map", "1:a:0", "-shortest", video_output
        ], check=True, capture_output=True)
        print(f"✅ Финальное видео успешно создано: {video_output}")
    except subprocess.CalledProcessError as e:
        print(f"❌ Ошибка FFmpeg: {e.stderr.decode()}")

# Заменяем аудио в оригинальном видео
print("--- Шаг: Финальная сборка видео ---")
if os.path.exists("final_audio.wav"):
    replace_audio_in_video(INPUT_VIDEO, "final_audio.wav", OUTPUT_VIDEO)
else:
    print("⚠️ Файл 'final_audio.wav' не обнаружен. Пожалуйста, запустите ячейку объединения аудио (0oVmIyplOtcO) сначала.")
```
